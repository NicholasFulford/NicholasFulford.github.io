# **Bell's Theorem: Hidden Variables vs. Inherent Indeterminism**

## **§1 - Background**

Consider the decay of a neutral of a pion into an electron positron pair:
$$\pi^0 \rightarrow e^+ + e^-$$
Spin is conserved in this decay and hence the electron and positron must have opposite spins.
This means that, if we measure the spin of the electron along some axis, then we instantly know the positron will have the opposite spin along that axis. Now, if the two particles are separated  in space prior to measurement, then a measurement of one particle seems to give information about the second particle no matter how far away it is. It was this idea of non-locality that upset Einstein, resulting in his famous remark: "Gott würfelt nicht". In order to try and come to terms with his, he favoured the so called 'hidden variables' interpretation, proposing that the spin of the particles is determined by nature but simply unknown to us. This was at odds with the more radical interpretation of Max Born & Niels Bohr who proposed that the spin of the particles is undetermined by nature itself prior to measurement. This debate could be finally settled thanks to a paper by John Bell which showed that these two interpretations would arrive at different statistical predictions for certain experiments.

### - Hidden Variables (Einstein)
The state of each particle is determined after the decay. That is, the outcome of a measurement of either particle along any axis is predetermined and our measurement simply reveals that outcome. Each time we make a measurement, the spin subsequently aligns with this axis and so we are not able to gain any more information about the spin along other axes. For this reason, we are not able to fully learn the components of the spin along each axis. The components of the spin along each axis are predetermined but simply hidden from us, hence the name 'hidden variables'.

### - Inherent Indeterminism (Born)
The spin of each particle is undetermined after the decay, not merely unknown to us but actually not-defined physically. The possible outcomes of a measurement along a particular axis follow a probability distribution described by the entangled quantum state of the two particles. We do not consider the spin of either particle to have any meaning prior to measurement, it is only after a measurement along a particular axis that the spin is determined to have some value along that axis.

### Distinguishing Between The Two
In order to test these two theories, consider Alice & Bob who are separated in space and have in their possession the electron & positron respectively. They can each measure the spin of their particle along any of three axes which lie in a plane, separated by $2\pi/3$. Label these axes $\hat a, \hat b, \hat c$. Suppose that Alice & Bob measure their particles along one of these axes independently at random. We are interested in the probability that Alice & Bob obtain the same measurement outcome, either both up or both down. It turns out that the two interpretations of Einstein and Born lead to different predictions for this probability.

## **§2 - Experimental Predictions**

## Hidden Variable Prediction
If we follow the hidden variable theory, then, when the positron and electron are created, their spins along each of the three axes are determined (but unknown to Alice & Bob). Since they must be opposite along each axis, there are 8 possible 'hidden' sets of spins that the particles can have:

|                        |          Electron           |          Positron           |
|:----------------------:|:---------------------------:|:---------------------------:|
|         $$1$$          | $$+\hat a,+\hat b,+\hat c$$ | $$-\hat a,-\hat b,-\hat c$$ |
|         $$2$$          | $$+\hat a,-\hat b,+\hat c$$ | $$-\hat a,+\hat b,-\hat c$$ |
|         $$3$$          | $$+\hat a,+\hat b,-\hat c$$ | $$-\hat a,-\hat b,+\hat c$$ |
|         $$4$$          | $$+\hat a,-\hat b,-\hat c$$ | $$-\hat a,+\hat b,+\hat c$$ |
|         $$5$$          | $$-\hat a,+\hat b,+\hat c$$ | $$+\hat a,-\hat b,-\hat c$$ |
|         $$6$$          | $$-\hat a,-\hat b,+\hat c$$ | $$+\hat a,+\hat b,-\hat c$$ |
|         $$7$$          | $$-\hat a,+\hat b,-\hat c$$ | $$+\hat a,-\hat b,+\hat c$$ |
|         $$8$$          | $$-\hat a,-\hat b,-\hat c$$ | $$+\hat a,+\hat b,+\hat c$$ |

Now Alice & Bob measure their particles along one of these axes independently at random. Let's work out the probability they obtain the same measurement outcome. For sets 1 & 8 this is zero, since the spins for the electron and positron are always opposite along any of the axes. For each of the remaining sets 2-7, there are 4 pairs of measurements for which Alice & Bob yield the same results. Looking at the second set for example, these are:

|  Alice's axis   |   Bob's axis    | Measurement |
|:---------------:|:---------------:|:-----------:|
|   $$\hat a$$    |   $$\hat b$$    |   $$++$$    |
|   $$\hat c$$    |   $$\hat b$$    |   $$++$$    |
|   $$\hat b$$    |   $$\hat a$$    |   $$--$$    |
|   $$\hat b$$    |   $$\hat c$$    |   $$--$$    |

and similarly for the other sets. There are 9 possible of pairs of axes that Alice & Bob could choose (aa,ab,ac,ba etc...) , hence the probability that they obtain the same measurement is $\frac{4}{9}$.
To not assume anything about the distribution of the 'hidden' outcome sets, suppose that the proportion of the particles produced in set $i$ is $p_i$. Hence, the probability of Alice & Bob yielding the same measurement result is:
$$P_{same} = \frac{4}{9}(p_1 + p_2 + p_3 + p_4 + p_5 + p_6 + p_7) \leq \frac{4}{9}$$
With equality iff $p_1 = p_8 = 0$.

## Indeterminism Prediction

To aid our set up, wlog let the axis that Alice chooses be $\hat z$. Then the state of the electron-positron pair prior to measurement can be written as:
$$|{\psi}\rangle = \frac{1}{\sqrt{2}}(|{\uparrow}\rangle_A|{\downarrow}\rangle_B + |{\downarrow}\rangle_A|{\uparrow}\rangle_B)$$
Where the kets $|{\uparrow}\rangle$ & $|{\downarrow}\rangle$ represent the state of a particle with spin up/down aligned along the $\hat z$ axis.

Now, Bob will measure the spin of his positron along one of the three axes, which are at an angle $\theta = 0, \frac{2\pi}{3}$ or $\frac{4\pi}{3}$ from $\hat{z}$ (Alice's axis). The spin operator (in the $|{\uparrow}\rangle$ & $|{\downarrow}\rangle$ basis) along an axis $\hat{n}$ with polar angle $\theta$ and azimuthal angle $\phi$ to $\hat z$ is:
$$\hat{S} = \frac{\hbar}{2}\begin{bmatrix} \cos{\theta} & e^{-i\phi}\sin{\theta} \\ e^{i\phi}\sin{\theta} & -\cos{\theta} \end{bmatrix}$$
In our case, $\phi$ is just the orientation about $\hat z$ of the plane containing the three axes, and turns out to not affect our results here. The eigenstates of this operator are:
$$|{\uparrow_{\hat{n}}}\rangle = \cos(\frac{\theta}{2})|{\uparrow}\rangle + e^{i\phi}\sin(\frac{\theta}{2})|{\downarrow}\rangle$$
$$|{\downarrow_{\hat{n}}}\rangle = \sin(\frac{\theta}{2})|{\uparrow}\rangle - e^{i\phi}\cos(\frac{\theta}{2})|{\downarrow}\rangle$$

Hence, once Bob has chosen one of the three axes to measure along, the probability that they both measure spin up is also:
$$P_{\uparrow\uparrow} = |(\langle{\uparrow}|_{A}\langle{\uparrow_{\hat{n}}}|_{B}) |{\psi}\rangle|^2$$
and following some algebra we obtain:
$$P_{\uparrow\uparrow} = \frac{1}{2}\sin^2(\frac{\theta}{2})$$
Similarly, the probability that they both measure spin down is:
$$P_{\downarrow\downarrow} = \frac{1}{2}\sin^2(\frac{\theta}{2})$$
Hence, given Bob has chosen a particular axis, the probability that they both obtain the same measurement outcome is:
$$P_{same}(\theta) = P_{\uparrow\uparrow} + P_{\downarrow\downarrow} = \sin^2(\frac{\theta}{2})$$

The three axes $\hat{a}, \hat{b}, \hat{c}$  Bob chooses from have a polar angle $\theta = 0, \frac{2\pi}{3}$ or $\frac{4\pi}{3}$ to $\hat z$ and he is equally likely to choose each of them. Hence the probability that Alice & Bob obtain the same measurement outcome after both choosing an axis independently at random is:
$$P_{same} =  \frac{1}{3}\sin^2(0) +  \frac{1}{3}\sin^2(\frac{\pi}{3}) +  \frac{1}{3}\sin^2(\frac{2\pi}{3})$$
$$ = 0 + \frac{1}{4} + \frac{1}{4}$$
$$= \frac{1}{2}$$

## **§3 - Testing on a Quantum Computer**

## Circuit set up:
These two predictions are incompatible since $\frac{1}{2}>\frac{4}{9}$. Hence, we should be able to distinguish between the two interpretations experimentally by carrying out the above process a large number of times. For that we will use a quantum computer. While we can't carry the exact decay described above, we can devise a quantum circuit that is equivalent to the decay process and consider the outcome statistics of measurements of this circuit on a quantum computer.

 We continue to define the axis Alice chooses to measure along as $+\hat z$ and wlog set $\phi = 0$. The qubits of our quantum computer start off initialised in the state $|{0}\rangle$ (which we identify with $|{\uparrow}\rangle$) and when measurements are made, they are along the $\hat z$ axis. Hence we must devise a quantum circuit such that a measurement of one of the qubits along $\hat z$ corresponds to (has the same probability distribution as) Bob measuring his positron along one of the three axes $\hat a,\hat b, \hat c$.

First off, we want our qubits that start initialised in the state $|{0}\rangle_A|{0}\rangle_B$ to be in the sate $\frac{1}{\sqrt{2}}(|{0}\rangle_A|{1}\rangle_B + |{1}\rangle_A|{0}\rangle_B)$. This can be done using $H, X$ and $CNOT$ gates like so:

$$(H\otimes I)|{0}\rangle_A|{0}\rangle_B = \frac{1}{\sqrt{2}}(|{0}\rangle_A|{0}\rangle_B + |{1}\rangle_A|{0}\rangle_B)$$
$$\Rightarrow (H\otimes X)|{0}\rangle_A|{0}\rangle_B = \frac{1}{\sqrt{2}}(|{0}\rangle_A|{1}\rangle_B + \|{1}\rangle_A|{1}\rangle_B)$$
$$\Rightarrow CNOT_{AB}(H\otimes X)|{0}\rangle_A|{0}\rangle_B = \frac{1}{\sqrt{2}}(|{0}\rangle_A|{1}\rangle_B + |{1}\rangle_A|{0}\rangle_B)$$

Here we use $CNOT_{AB}$ to mean that $A$ is the control and $B$ is the target.

Our quantum computer only measures along $\hat z$, however, Bob is free to measure along any of the three axes $\hat a,\hat b, \hat c$ - two of which are at an angle $\theta = \frac{2\pi}{3}$ or $\frac{4\pi}{3}$ to $\hat z$. Hence, we must apply the relevant gates such that a final measurement of the second qubit (Bob's) along $\hat z$ is equivalent (probabilistically speaking) to Bob measuring  $\frac{1}{\sqrt{2}}(|{0}\rangle_A|{1}\rangle_B + |{1}\rangle_A|{0}\rangle_B)$ along $\hat a,\hat b$ or $\hat c$.

Bob is measuring along an axis $\hat n$ which is at an angle $\theta$ to $\hat z$. This is equivalent to measuring the second qubit along $\hat z$ if we rotate this qubit by an angle $-\theta$ to $\hat z$ prior to measurement. For this we need a rotation operator and since we set $\phi=0$, we are rotating about the $\hat y$ axis. The rotation operator along the $\hat y$ axis for a single qubit is:
$$R_y(\theta)=\begin{bmatrix} \cos(\theta/2) & -\sin(\theta/2) \\ \sin(\theta/2) & \cos(\theta/2) \end{bmatrix}$$

Hence, we apply $R_y(-\theta)$ to the second qubit to obtain our final state prior to measurement:
$$(I\otimes R_y(-\theta))CNOT_{AB}(H\otimes X)|{0}\rangle_A|{0}\rangle_B$$

Bob is choosing between the axes independently at random. Hence, doing this a large number of times, Bob will choose each of $\hat a,\hat b$ or $\hat c$ with equal proportion $\frac{1}{3}$. Hence, we will measure each of the three circuits an equal number of times and then consider the combined results.

## Making measurements:

Now we are finally ready to carry out our experiment. We will take an equal number of measurement outcomes for each circuit and look at the total number of times the outcomes for both qubits were the same.  Remember that, for our circuits, the first qubit corresponds to Alice's electron and the second qubit corresponds to Bob's positron. Furthermore, a measurement outcome of $0$ or $1$ correspond to a measurement of spin $\uparrow$ or $\downarrow$ respectively.

First, let us construct the three circuits:

```python
from qiskit import QuantumCircuit
from numpy import pi

qcs = [QuantumCircuit(2, 2), QuantumCircuit(2, 2), QuantumCircuit(2, 2)]
for i in range(0, len(qcs)):
    qcs[i].h(0)
    qcs[i].x(1)
    qcs[i].cx(0, 1)

qcs[1].ry(-2 * pi / 3, 1)
qcs[2].ry(-4 * pi / 3, 1)

for i in range(0, len(qcs)):
    qcs[i].barrier()
    qcs[i].measure([0, 1], [0, 1])

counts_list = [None] * len(qcs)
```

```python
qcs[0].draw(output="mpl")
```

#### Circuit 2  (Rotation by $\theta = -{2\pi}/{3}$)

```python
qcs[1].draw(output="mpl")
```

#### Circuit 3  (Rotation by $\theta = -{4\pi}/{3}$)

```python
qcs[2].draw(output="mpl")
```

## First, a simulation

To check our code runs as intended, we first run it on a simulator. In this case we will take a sample of 10,000 measurements from each circuit.

```python
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram
sampler = StatevectorSampler()
# For each circuit we take a sample of 10,000 measurements
# Here, real_counts_list is a list of dictionaries for each circuit storing each outcome count
for i in range(0, len(qcs)):
    job = sampler.run([qcs[i]], shots=10000)
    result = job.result()
    data_pub = result[0].data
    counts = data_pub.c.get_counts()
    counts_list[i] = counts

outcomes = ("00", "01", "10", "11")
# Converting "None"s in real_counts_list into 0's so that we can sum counts.
for i in range(0, len(qcs)):
    for j in range(0, len(outcomes)):
        if counts_list[i].get(outcomes[j]) is None:
            counts_list[i].update({outcomes[j]: 0})

# Creating a dictionary that stores the total outcome count for all three circuits
total_counts = {}
for i in range(0, len(outcomes)):
    total_counts[outcomes[i]] = sum(counts_list[j].get(outcomes[i]) for j in range(0, len(qcs)))
print("Outcome counts")
for i in range(len(outcomes)):
    print(f"'{outcomes[i]}' - {total_counts[outcomes[i]]} ")
plot_histogram(total_counts)
```

As we can see from the above histogram, each possible measurement outcome pair appears with roughly equal proportion, indicating that the probability of Alice & Bob obtaining the same outcome ($00$ or $11$) is $\frac{1}{2}$. This is, however, just a simulation of the underlying quantum mechanics and so doesn't prove anything just yet.

## Running on real hardware

Before we can run these circuits on a real quantum computer, we will need to convert them into an instruction set architecture (ISA). This is so that they can be run on the particular hardware we are using, which in this case will be IBM Brisbane. We will be using Qiskit's preset pass manager which does this transpilation process automatically.

```python
from qiskit_ibm_runtime import QiskitRuntimeService
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime import SamplerV2 as Sampler
from qiskit_ibm_runtime import Batch

service = QiskitRuntimeService()
backend = service.backend("ibm_brisbane")
pm = generate_preset_pass_manager(optimization_level=3, backend=backend)
# Converting our circuits into ISA circuits for the hardware we are using (IBM Brisbane)
qcs_isa = qcs
for i in range(0, len(qcs)):
    qcs_isa[i] = pm.run(qcs[i])
```

```python
qcs_isa[0].draw(output="mpl", idle_wires=False, style="iqp")
```

#### Circuit 2  (Rotation by $\theta = -{2\pi}/{3}$) ISA

```python
qcs_isa[1].draw(output="mpl", idle_wires=False, style="iqp")
```

```python
qcs_isa[2].draw(output="mpl", idle_wires=False, style="iqp")
```

## Final results

```python
batch = Batch(backend=backend)
sampler = Sampler(mode=batch)

result_list = [None] * len(qcs)
real_counts_list = [None] * len(qcs)
# For each circuit we take a sample of 10,000 measurements using the sampler primitive
# Here, real_counts_list is a list of dictionaries that store the outcome counts for each circuit
for i in range(0, len(qcs)):
    job = sampler.run([qcs_isa[i]], shots=10000)
    result_list[i] = job.result()
    data_pub = result_list[i][0].data
    counts = data_pub.c.get_counts()
    real_counts_list[i] = counts
batch.close()

outcomes = ("00", "01", "10", "11")
# Converting "None"s in real_counts_list into 0's so that we can sum counts.
for i in range(0, len(qcs)):
    for j in range(0, len(outcomes)):
        if real_counts_list[i].get(outcomes[j]) is None:
            real_counts_list[i].update({outcomes[j]: 0})

# Creating a dictionary that stores the total outcome count for all three circuits
real_total_counts = {}
for i in range(0, len(outcomes)):
    real_total_counts[outcomes[i]] = (
        sum(real_counts_list[j].get(outcomes[i]) for j in range(0, len(qcs)))
    )
print("Outcome counts")
for i in range(len(outcomes)):
    print(f"'{outcomes[i]}' - {real_total_counts[outcomes[i]]} ")
plot_histogram(real_total_counts)
```

Above is an illustration of the output from the real quantum computer. Running the code multiple times, we consistently obtain the same output - that is, each possible two-qubit measurement outcome appearing in roughly equal proportion. From the above data, the proportion of equal outcomes ($00$ or $11$) is:
$$P_{same} = \frac{7356+7639}{30000} = 0.4998 ...$$
which is in agreement with the inherent indeterminism prediction of $P_{same} = \frac{1}{2}$. Indeed, running the code multiple times, we consistently obtain values of $P_{same} ≈ 0.5$.

Let's run over what we've just shown here. We started with the decay $\pi^0 \rightarrow e^+ + e^-$ and then considered Alice & Bob who measure the spin of each particle independently at random along one of three axes $\hat a, \hat b,\hat c$. We then considered the probability of both of them obtaining the same measurement outcome. This was important since the proposed interpretations of Einstein and Born lead to different predictions for this probability $P_{same}$. The next step was to translate this scenario onto a quantum circuit which could be run on a quantum computer, which we did (using three circuits in the end). The final results from our quantum computer indicate that the probability of Alice & Bob obtaining the same measurement outcome is indeed $1/2$, refuting the hidden variables interpretation and providing evidence for the true non-deterministic nature of quantum mechanics. This means that, after the decay, the spins are not in any well-defined state and the particles carry no hidden variable instructions with them to determine the outcomes of future measurements. However, the spins will always be opposite when measured along the same axis! Somehow, the physics of one particle is determined by what is done to the other, regardless of how far away it is. This, is Einstein's "spooky action at a distance" and is a perfect example of why quantum mechanics is so fascinating.
